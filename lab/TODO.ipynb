```python
def make_todos(d: dict) -> str:
    output = "# 🚧 Things to Do" + NEWLINE * 2
    for file_name, strings in d.items():
        output += f"## `{file_name}`\n"

        for s in strings:
            line_number, _, todo_str = s.split(COLON, maxsplit=2)
                
            output += f"* **Line {line_number}**: {todo_str}\n"
    pyperclip.copy(output)
    return output
```

```python
display(Markdown(make_todos(TODOS)))
```

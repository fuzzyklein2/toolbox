```python
def load(p: Path) -> dict:
    with p.open() as f:
        d = json.load(f)
    return d
```

```python
def dump(d: dict, p: Path):
    with p.open('w') as f:
        json.dump(d, f)
```

```python
load(STRUM_CONFIG_FILE)
```

```python
def config(key: str, value: str) -> dict:
    d = load(STRUM_CONFIG_FILE)
    d[key] = value
    dump(d)
    return d
```

```python
def dump(d: dict, path: Path) -> dict:
    with path.open('w') as f:
        json.dump(d, f)
```

```python
def configure(key: str, value: str, path: Path) -> dict:
    """ Add `key : value` to the JSON file at path. """
    if not path.exists() or len(path.read_text()) == 0:
        path.write_text('{}')
    
    d = load(path)
    d[key] = value
    d = dump(d, path)
```

```python
configure('author', "Russell Alan Klein", STRUM_CONFIG_FILE)
```
